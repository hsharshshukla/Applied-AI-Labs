Remove html tags

```python
import re
def remove_html_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'',text)
```

Remove url

```python
def remove_url(text):
  pattern = re.compile(r'https?://\S+|www\.\S')
  return pattern.sub(r'',text)
```

```python
text = 'Check on my youtube channel https://www.youtube.com'
text = remove_url(text)
print(text)
```

Punctuation Handling

```python
import string,time
string.punctuation
```

```python
exclude  = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char,'')
  return text
```

```python
text = 'string. With. Punctuation?'
```

```python
start = time.time()
print(remove_punc(text))
time1 = time.time() - start
print(time1 * 50000)
```

```python
def remove_punc1(text):
  return text.translate(str.maketrans('','',exclude))
```

```python
start = time.time()
remove_punc1(text)
time2 = time.time() -start
print(time2 * 50000)
```

Chat Conversation Handle

```python
chat_words = {'FYI':'For your Information',
              'ASAP':'As soon as possible'}
```

```python
def chat_conversion(text):
  new_text = []
  for w in text.split():
    if w.upper() in chat_words:
      new_text.append(chat_words[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```

```python
chat_conversion('Do this work ASAP')
```

StopWords

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
len(stopwords.words('english'))
```

```python
def remove_stopwords(text):
  new_text = []

  for word in text.split():
    if word in stopwords.words('english'):
      new_text.append('')
    else:
      new_text.append(word)
  x = new_text[:]
  new_text.clear()
  return " ".join(x)
```

Remove Emoji

```python
import re
def remove_emoji(text):
  emoji_pattern = re.compile("["
    u"\U0001F600-\U0001F64F"
    u"\U0001F300-\U0001F5FF"
    u"\U0001F680-\U0001F6FF"
    u"\U0001F1E0-\U0001F1FF"
    u"\U00002702-\U000027B0"
    u"\U000024C2-\U0001F251"
    "]+",  flags=re.UNICODE)
  return emoji_pattern.sub(r'',text)
```

**Stemmer**  - stemmed words are not always readable



```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
  return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

### **Lemmatization** - Readable & slow as compared to Stemming

```python
import nltk
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "he was running and eating at same time. He has bad habit of swimming after playing long hours in the"

punctuations = "?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
  if word in punctuations:
    sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))

for word in sentence_words:
  print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
```
