Function Calling in OpenAI  - this is to avoid manual data extraction and then feeding into model, instead with function calling model can directng interact with response data.

```python
import os
import openai
from dotenv import load_dotenv
```

```python
load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
openai.api_key = OPENAI_API_KEY
```

```python
prompt = " Hello how are you ?"

response = openai.ChatCompletion.create(
    model = "gpt-3.5-turbo",
    messages=[
        {"role":"system","content":"You are a helpful assistant"},
        {"role":"user","content":prompt}
    ]
)

print(response)
```
